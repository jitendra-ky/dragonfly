```python
# here I will test the registration flow
import requests
from pprint import pprint

base_url = "http://localhost:8000/"

# health check
res = requests.get(base_url + "api/health/")

print(res.status_code, res.reason)
```

```python
# initiate registration request
res = requests.post(
    base_url + "api/user-profile/",
    json= {
        "contact" : "this is contact",
        "email" : "test3@gmail.com",
        "password" : "rootroot"
    }
)

print(res.status_code, res.reason)
```

```python
## activate the user with OPT
res = requests.post(
    base_url + "api/sign-up-otp/",
    json= {
        "email" : "test3@gmail.com",
        "otp" : 652054
    }
)

print(res.status_code, res.reason)
```

```python
# try to get the user JWT tokens
res = requests.post(
    base_url + "api/sign-in/",
    json={
        "email": "test3@gmail.com",
        "password": "rootroot"
    }
)

print(res.status_code, res.reason)
print(res.json())

access_token = res.json().get("access")
print(access_token)
```

```python
# check are we logged in
res = requests.get(
    base_url + "api/sign-in/",
    headers= {
        "Authorization" : f"Bearer {access_token}",
        "Content-Type" : "application/json",
        },
)

print(res.status_code, res.reason)

if (res.status_code < 400):
    pprint(res.json())
```
